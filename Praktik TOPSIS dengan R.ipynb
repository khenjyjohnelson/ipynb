```r
# METODE TOPSIS
# TANPA KRITERIA BENEFIT DAN KRITERIA COST

# BOBOT
w <- c(20, 20, 30, 30)

# ALTERNATIF
# Alternatif 1: Andi
a1 <- c(2, 1, 4, 5)
# Alternatif 2: Budi
a2 <- c(1, 4, 2, 3)
# Alternatif 3: Charles
a3 <- c(3, 2, 3, 3)
# Alternatif 4: Dani
a4 <- c(2, 4, 5, 3)
# Alternatif 5: Edward
a5 <- c(3, 2, 2, 3)
# Alternatif 6: Fanny
a6 <- c(4, 1, 1, 1)
# Alternatif 7: Garfield
a7 <- c(5, 3, 5, 4)

# MATRIKS KEPUTUSAN
x <- data.frame(a1, a2, a3, a4, a5, a6, a7)
x <- data.frame(t(x))



#SET KRITERIA BENEFIT DAN KRITERIA COST
c <- c('benefit', 'benefit', 'benefit', 'cost')


# NORMALISASI MATRIKS KEPUTUSAN
n_row <- nrow(x)
n_col <- ncol(x)
akar.total.kuadrat <- c()
r <- data.frame(matrix(nrow = n_row, ncol = n_col))

for (i in 1:n_col){
  for (j in 1:n_col) {
    akar.total.kuadrat[j] <- sqrt(sum(x[j]^2))
    r[, j] <- x[, j] / akar.total.kuadrat[j]
  }
}

print(r)


# MATRIKS KEPUTUSAN YANG TERNORMALISASI TERBOBOT
y <- data.frame(matrix(nrow = n_row, ncol = n_col))

for (i in 1:n_col) {
    for (j in 1:n_col) {
        y[, j] <- w[j] * r[, j]
    }
}

print(y)

# MATRIKS SOLUSI IDEAL POSITIF dan NEGATIF

y.plus <- c()
y.minus <- c()

for (i in 1:n_row) {
    for (j in 1:n_col) {
        if (c[j] == 'benefit') {
            y.plus[j] <- max(y[, j])
            y.minus[j] <- min(y[, j])
          } else if (c[j] == 'cost') {
            y.plus[j] <- min(y[, j])
            y.minus[j] <- max(y[, j])
          }
    }
}

A.plus <- y.plus
A.minus <- y.minus

print(A.plus)
print(A.minus)




# JARAK ANTARA NILAI ALTERNATIF DAN MATRIKS IDEAL POSITIF
temp_d <- data.frame(matrix(nrow = n_row, ncol = n_col))
D.plus <- c()

for (i in 1:n_row) {
  for (j in 1:n_col) {
    temp_d[i, j] <- (y[i, j] - y.plus[j])^2
    D.plus[i] <- sqrt(sum(temp_d[i, ]))
  }
}

print(D.plus)


# JARAK ANTARA NILAI ALTERNATIF DAN MATRIKS IDEAL NEGATIF
temp_d <- data.frame(matrix(nrow = n_row, ncol = n_col))
D.minus <- c()

for (i in 1:n_row) {
  for (j in 1:n_col) {
    temp_d[i, j] <- (y[i, j] - y.minus[j])^2
    D.minus[i] <- sqrt(sum(temp_d[i,]))
  }
}

print(D.minus)


# NILAI PREFERENSI SETIAP ALTERNATIF
v <- c()
for (i in 1:n_row) {
  v[i] <- D.minus[i] / (D.minus[i] + D.plus[i])
}

print(v)


# PERANGKINGAN TERHADAP ALTERNATIF
for (i in 1:n_row) {
  if (v[i] == max(v)) {
    cat("Alternatif yang dipilih adalah Alternatif", i)
  }
}
```
