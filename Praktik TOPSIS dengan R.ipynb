# Praktikum Latihan Vava Vivi Vovo

```r
# METODE TOPSIS
# TANPA KRITERIA BENEFIT DAN KRITERIA COST

# BOBOT
w <- c(5, 3, 2)

# ALTERNATIF
# Alternatif 1: Vava
a1 <- c(2, 3, 5)
# Alternatif 2: Vivi
a2 <- c(5, 1 ,3)
# Alternatif 3: Vovo
a3 <- c(1, 1, 5)

# MATRIKS KEPUTUSAN
x <- data.frame(a1, a2, a3)
x <- data.frame(t(x))
```

# All

```r
# NORMALISASI MATRIKS KEPUTUSAN
n_row <- nrow(x)
n_col <- ncol(x)
akar.total.kuadrat <- c()
r <- data.frame(matrix(nrow = n_row, ncol = n_col))

for (i in 1:n_col){
  for (j in 1:n_col) {
    akar.total.kuadrat[j] <- sqrt(sum(x[j]^2))
    r[, j] <- x[, j] / akar.total.kuadrat[j]
  }
}

print(r)
```

```r
# MATRIKS KEPUTUSAN YANG TERNORMALISASI TERBOBOT
y <- data.frame(matrix(nrow = n_row, ncol = n_col))

for (i in 1:n_col) {
    for (j in 1:n_col) {
        y[, j] <- w[j] * r[, j]
    }
}

print(y)
```

# Praktikum Contoh 1

```r
# MATRIKS SOLUSI IDEAL POSITIF
y.plus <- c()
for (j in 1:n_col) {
  y.plus[j] <- max(y[, j])
}
A.plus <- y.plus

print(A.plus)
```

```r
# MATRIKS SOLUSI IDEAL NEGATIF
y.minus <- c()
for (j in 1:n_col) {
  y.minus[j] <- min(y[, j])
}
A.minus <- y.minus

print(A.minus)
```

# Praktikum Contoh 2

```r
# MATRIKS SOLUSI IDEAL POSITIF dan NEGATIF

y.plus <- c()
y.minus <- c()

for (i in 1:n_row) {
    for (j in 1:n_col) {
        if (c[j] == 'benefit') {
            y.plus[j] <- max(y[, j])
            y.minus[j] <- min(y[, j])
          } else if (c[j] == 'cost') {
            y.plus[j] <- min(y[, j])
            y.minus[j] <- max(y[, j])
          }
    }
}

A.plus <- y.plus
A.minus <- y.minus

print(A.plus)
print(A.minus)
```

# All

```r
# JARAK ANTARA NILAI ALTERNATIF DAN MATRIKS IDEAL POSITIF
temp_d <- data.frame(matrix(nrow = n_row, ncol = n_col))
D.plus <- c()

for (i in 1:n_row) {
  for (j in 1:n_col) {
    temp_d[i, j] <- (y[i, j] - y.plus[j])^2
    D.plus[i] <- sqrt(sum(temp_d[i, ]))
  }
}

print(D.plus)
```

```r
# JARAK ANTARA NILAI ALTERNATIF DAN MATRIKS IDEAL NEGATIF
temp_d <- data.frame(matrix(nrow = n_row, ncol = n_col))
D.minus <- c()

for (i in 1:n_row) {
  for (j in 1:n_col) {
    temp_d[i, j] <- (y[i, j] - y.minus[j])^2
    D.minus[i] <- sqrt(sum(temp_d[i,]))
  }
}

print(D.minus)
```

```r
# NILAI PREFERENSI SETIAP ALTERNATIF
v <- c()
for (i in 1:n_row) {
  v[i] <- D.minus[i] / (D.minus[i] + D.plus[i])
}

print(v)
```

```r
# PERANGKINGAN TERHADAP ALTERNATIF
for (i in 1:n_row) {
  if (v[i] == max(v)) {
    cat("Alternatif yang dipilih adalah Alternatif", i)
  }
}
```
