# 2. All

```r
# Iterasi dan Perhitungan Eigen
# Fungsi iterasi eigenvector
eigen_iterasi <- function(mat, threshold = 0.01, max_iter = 100){
    v <- rep(1, nrow(mat))
    v <- v / sum(v)
    cat("Iterasi ke-0:", paste(round(v, 4), collapse = " "), "\n")
    
    for(i in 1:max_iter){
        v_new <- mat %*% v
        v_new <- v_new / sum(v_new)
        delta <- max(abs(v_new - v) )
        cat(sprintf("Iterasi ke-%2d : %s | A = %.6f\n", i, paste(round(v_new, 4), collapse = " "), delta))
        if(delta < threshold){
            cat(sprintf("Konvergen pada iterasi ke-%2d\n\n", i) )
            return(round (v_new, 4) )
        }
    V <- v_new
    }
    cat("Maksimum iterasi tercapai\n\n")
    return(round(v, 4))
}
```

# 3a. Soal Komputer

```r
cat("=== ITERASI: KRITERIA ===\n")
eigen_kriteria <- eigen_iterasi(kriteria_matrix)

cat("=== ITERASI: KOMPABILITAS ===\n")
eigen_kompabilitas <- eigen_iterasi(mat_kompabilitas)

cat("=== ITERASI: UNJUK KERJA ===\n")
eigen_unjuk <- eigen_iterasi(mat_unjuk)

cat("=== ITERASI: PURNA JUAL ===\n")
eigen_purna <- eigen_iterasi(mat_purna)

# Matriks Alternatif Gabungan
alt_matrix <- cbind(eigen_kompabilitas, eigen_unjuk, eigen_purna)
colnames(alt_matrix) <- c("Kompabilitas", "Unjuk Kerja", "Purna Jual")
rownames(alt_matrix) <- c("IBM", "Acer", "HP", "Compaq")
```

```r
# Tampilkan Hasil
cat("\n === BOBOT KRITERIA === \n")
print(eigen_kriteria)
#cat("\nCR Kriteria:", cr kriteria$CR, "-", cr kriteria$status, "\n")

cat("\n === BOBOT ALTERNATIF per KRITERIA === \n")
alt_matrix <- cbind(eigen_kompabilitas, eigen_unjuk, eigen_purna)
colnames(alt_matrix) <- c("Kompabilitas", "Unjuk Kerja", "Purna Jual")
rownames(alt_matrix) <- c("IBM", "Acer", "HP", "Compaq")


print(round(alt_matrix, 4))

# Pembobotan dan Perangkingan Alternatif
cat("\n === HASIL PENJUMLAHAN DAN PERANGKINGAN ALTERNATIF === \n")

# Matriks pembobotan (global weight = local weight * kriteria weight)
bobot_global <- sweep(alt_matrix, 2, eigen_kriteria, "*") # perkalian kolom
bobot_global <- round(bobot_global, 4)

# Tambahkan skor total (jumlah baris)
skor_akhir <- rowSums (bobot_global)
bobot_global <- cbind(bobot_global, Total = round(skor_akhir, 4))
```

```r
# Fungsi menghitung CR
hitung_CR <- function (mat, eigenvec) {
    n <- nrow(mat)
    wsv <- mat %*% eigenvec
    cv <- wsv / eigenvec
    lambda_max <- mean (cv)
    CI <- (lambda_max - n) / (n - 1)
    RI_values <- c(0.00, 0.58, 0.90, 1.12, 1.24, 1.32, 1.41, 1.45, 1.49)
    RI <- ifelse(n <= length(RI_values), RI_values[n], NA)
    CR <- CI / RI
    return(list(CI = round(CI, 4), CR = round(CR, 4),
    status = ifelse(CR <= 0.1, "Konsisten", ifelse(CR <= 0.15, "Agak Konsisten", "Tidak Konsisten"))))
}



# Perhitungan CR
cr_kriteria <- hitung_CR(kriteria_matrix, eigen_kriteria)
cr_kompatibilitas <- hitung_CR(mat_kompabilitas, eigen_kompabilitas)
cr_unjuk <- hitung_CR(mat_unjuk, eigen_unjuk)
cr_purna <- hitung_CR(mat_purna, eigen_purna)


#Tampilkan Hasil CR
cat("\nCR Kompatibilitas:", cr_kompatibilitas$CR, "-", cr_kompatibilitas$status, "\n")
cat("CR Unjuk Kerja:", cr_unjuk$CR, "-", cr_unjuk$status, "\n")
cat("CR Purna Jual:", cr_purna$CR, "-", cr_purna$status, "\n")
```

```r
# Fungsi menghitung CR
hitung_CR <- function (mat, eigenvec) {
    n <- nrow(mat)
    wsv <- mat %*% eigenvec
    cv <- wsv / eigenvec
    lambda_max <- mean(cv)
    CI <- (lambda_max - n) / (n - 1)
    RI_values <- c(0.00, 0.58, 0.90, 1.12, 1.24, 1.32, 1.41, 1.45, 1.49)
    RI <- ifelse(n <= length(RI_values), RI_values[n], NA)
    CR <- CI / RI
    return(list(CI = round(CI, 4), CR = round(CR, 4),
        status = ifelse(CR <= 0.1, "Konsisten", ifelse(CR <= 0.15, "Agak Konsisten", "Tidak Konsisten"))))
}

# Perhitungan CR
cr_kriteria <- hitung_CR(kriteria_matrix, eigen_kriteria)
cr_cantik <- hitung_CR(mat_cantik, eigen_cantik)
cr_baik <- hitung_CR(mat_baik, eigen_baik)
cr_pintar <- hitung_CR(mat_pintar, eigen_pintar)
cr_taat <- hitung_CR(mat_taat, eigen_taat)
cr_kaya <- hitung_CR(mat_kaya, eigen_kaya)

# Tampilkan Hasil CR
cat("\nCR Cantik:", cr_cantik$CR, "-", cr_cantik$status, "\n")
cat("CR Baik:", cr_baik$CR, "-", cr_baik$status, "\n")
cat("CR Pintar:", cr_pintar$CR, "-", cr_pintar$status, "\n")
cat("CR Taat Agama:", cr_taat$CR, "-", cr_taat$status, "\n")
cat("CR Kaya:", cr_kaya$CR, "-", cr_kaya$status, "\n")
```
