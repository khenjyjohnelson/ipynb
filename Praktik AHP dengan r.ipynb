```r
kriteria_matrix <- matrix(c(
    1, 0.5, 3,
    2, 1, 4,
    1/3, 1/4, 1
), nrow = 3, byrow = TRUE)
colnames(kriteria_matrix) <- rownames (kriteria_matrix) <- c("Kompabilitas", "Unjuk Kerja", "Purna Jual")


mat_kompabilitas <- matrix (c(
    1, 0.25, 4, 0.167,
    4, 1, 4, 0.25,
    0.25, 0.25, 1, 0.2,
    6, 4, 5, 1
), nrow = 4, byrow = TRUE)

mat_unjuk <- matrix (c(
    1, 2, 5, 1,
    0.5, 1, 3, 2,
    0.2, 0.333, 1, 0.25,
    1, 0.5, 4, 1
), nrow = 4, byrow = TRUE)

mat_purna <- matrix (c(
    1, 4, 5, 2,
    1/3, 1, 2, 3,
    0.25, 0.5, 1, 0.333,
    1, 2, 2, 1
), nrow = 4, byrow = TRUE)

rownames(mat_kompabilitas) <- rownames(mat_unjuk) <-
rownames(mat_purna) <- colnames(mat_kompabilitas) <- colnames(mat_unjuk) <-
colnames(mat_purna) <- c("IBM", "Acer", "HP", "Compaq")
```

```r
# Iterasi dan Perhitungan Eigen
# Fungsi iterasi eigenvector
eigen_iterasi <- function(mat, threshold = 0.01, max_iter = 100){
    v <- rep(1, nrow(mat))
    v <- v / sum(v)
    cat("Iterasi ke-0:", paste(round(v, 4), collapse = " "), "\n")
    
    for(i in 1:max_iter){
        v_new <- mat %*% v
        v_new <- v_new / sum(v_new)
        delta <- max(abs(v_new - v) )
        cat(sprintf("Iterasi ke-%2d : %s | A = %.6f\n", i, paste(round(v_new, 4), collapse = " "), delta))
        if(delta < threshold){
            cat(sprintf("Konvergen pada iterasi ke-%2d\n\n", i) )
            return(round (v_new, 4) )
        }
    V <- v_new
    }
    cat("Maksimum iterasi tercapai\n\n")
    return(round(v, 4))
}
```

```r
cat("=== ITERASI: KRITERIA ===\n")
eigen_kriteria <- eigen_iterasi(kriteria_matrix)

cat("=== ITERASI: KOMPABILITAS ===\n")
eigen_kompabilitas <- eigen_iterasi(mat_kompabilitas)

cat("=== ITERASI: UNJUK KERJA ===\n")
eigen_unjuk <- eigen_iterasi(mat_unjuk)

cat("=== ITERASI: PURNA JUAL ===\n")
eigen_purna <- eigen_iterasi(mat_purna)
```

```r
# Tampilkan Hasil
cat("\n === BOBOT KRITERIA === \n")
print(eigen_kriteria)
#cat("\nCR Kriteria:", cr kriteria$CR, "-", cr kriteria$status, "\n")

cat("\n === BOBOT ALTERNATIF per KRITERIA === \n")
alt_matrix <- cbind(eigen_kompabilitas, eigen_unjuk, eigen_purna)
colnames(alt_matrix) <- c("Kompabilitas", "Unjuk Kerja", "Purna Jual")
rownames(alt_matrix) <- c("IBM", "Acer", "HP", "Compaq")


print(round(alt_matrix, 4))

# Pembobotan dan Perangkingan Alternatif
cat("\n === HASIL PENJUMLAHAN DAN PERANGKINGAN ALTERNATIF === \n")

# Matriks pembobotan (global weight = local weight * kriteria weight)
bobot_global <- sweep(alt_matrix, 2, eigen_kriteria, "*") # perkalian kolom
bobot_global <- round(bobot_global, 4)

# Tambahkan skor total (jumlah baris)
skor_akhir <- rowSums (bobot_global)
bobot_global <- cbind(bobot_global, Total = round(skor_akhir, 4))
```

```r
# Urutkan berdasarkan total skor
peringkat <- order(-bobot_global[, "Total"])
bobot_global_sorted <- bobot_global[peringkat, , drop = FALSE]

# Tampilkan tabel akhir pembobotan dan peringkat
print(bobot_global_sorted)

# Tampilkan peringkat secara eksplisit
cat("\n === PERINGKAT ALTERNATIF === \n")
for (i in 1:nrow(bobot_global_sorted) ) {
    cat(sprintf("%d. %s : %.4f\n", i, rownames(bobot_global_sorted) [i], bobot_global_sorted[i, "Total"] ) )
}
```
